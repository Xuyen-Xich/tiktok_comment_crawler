# TikTok Comments Classroom Demo

This notebook shows a simple social listening workflow: crawl comments, load them into pandas, clean text, score basic sentiment, find top keywords, and visualize engagement.

## 1. Crawl TikTok Comments

Run this command in a terminal first. Keep the browser visible so you can solve TikTok verification if it appears.

```powershell
python app.py crawl --url "https://www.tiktok.com/@creator/video/1234567890" --max-comments 200 --format all
```

````python
from pathlib import Path
import pandas as pd

output_dir = Path('../output')
json_files = sorted(output_dir.glob('*.json'))
json_files[-5:]
````

````python
df = pd.read_json(json_files[-1])
df.head()
````

## 2. Clean Text

````python
import sys
sys.path.append('..')

from analytics.preprocessing import normalize_comment_text, top_keywords
from analytics.sentiment import label_sentiment, score_sentiment

df['clean_text'] = df['comment_text'].fillna('').apply(normalize_comment_text)
df[['comment_text', 'clean_text']].head()
````

## 3. Basic Sentiment Analysis

````python
df['sentiment_score'] = df['comment_text'].fillna('').apply(score_sentiment)
df['sentiment_label'] = df['comment_text'].fillna('').apply(label_sentiment)
df['sentiment_label'].value_counts()
````

## 4. Top Keywords

````python
keywords = top_keywords(df['comment_text'].fillna(''), limit=20)
pd.DataFrame(keywords, columns=['keyword', 'count'])
````

## 5. Engagement Analysis

````python
engagement = df[['username', 'comment_text', 'like_count', 'sentiment_label']].sort_values('like_count', ascending=False)
engagement.head(10)
````

## 6. Visualization

````python
import matplotlib.pyplot as plt

sentiment_counts = df['sentiment_label'].value_counts()
sentiment_counts.plot(kind='bar', title='Comment Sentiment')
plt.xlabel('Sentiment')
plt.ylabel('Comments')
plt.show()
````
